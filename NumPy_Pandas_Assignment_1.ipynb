## Numpy Array Operations

```python
import numpy as np
import pandas as pd
```

```python
print("NumPy and Pandas are ready!")
```

### Create a 1D array of Week 1 temperatures

```python
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print(temperatures_w1)
```

### Check array shape, data type and number of elements

```python
print("Shape:", temperatures_w1.shape)
print("Data type:", temperatures_w1.dtype)
print("Number of elements:", temperatures_w1.size)
```

### Convert Celsius to Fahrenheit and calculate maximum, minimum and mean

```python
temperatures_f = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:", temperatures_f)
print("Maximum temperature:", temperatures_f.max())
print("Minimum temperature:", temperatures_f.min())
print("Mean temperature:", temperatures_f.mean())
```

### Perform array slicing and indexing

```python
print("First 3 days:", temperatures_w1[:3])

print("Weekend temperatures:", temperatures_w1[-2:])

print("Middle 3 days:", temperatures_w1[2:5])
```

### Create a 2D array of Week 1 and Week 2 temperatures

```python
temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print(temperatures)
```

### Inspect 2D array and extract temperatures

```python
print("Shape:", temperatures.shape)
print("Data type:", temperatures.dtype)
print("Total number of elements:", temperatures.size)

print("Week 1 temperatures:", temperatures[0])
print("Week 2 temperatures:", temperatures[1])

print("Week 1 weekend:", temperatures[0, -2:])
print("Week 2 weekend:", temperatures[1, -2:])
```

## Pandas Series

### Creating Pandas Series

```python
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=["Rank1", "Rank2", "Rank3", "Rank4", "Rank5"]
)

print(marks)
```

 ### Mark of the 1st rank student

```python
print("First rank mark:", marks.iloc[0])
```

### Marks of the top 3 ranks

```python
print("Top 3 ranks:", marks.loc["Rank1":"Rank3"])
```

 ### Mark of the 3rd rank student

```python
print("Third rank mark:", marks.iloc[2])
```

 ### Ranks where the marks are greater than 90.

```python
print("Marks above 90:", marks[marks > 90])
```

### Modify Rank1 mark

```python
marks.loc["Rank1"] = 100

print("Updated marks:")
print(marks)
```

### Remove Rank5

```python
marks = marks.drop("Rank5")

print("Updated marks:")
print(marks)
```

### Calculate CGPA

```python
cgpa = marks / 10

print("CGPA:")
print(cgpa)
```

## Pandas DataFrame 


### Creating Pandas DataFrame

```python
data = {
    "TransactionID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "ProductCategory": ["Electronics", "Clothing", "Electronics", "Furniture", "Clothing", "Electronics", "Furniture", "Clothing", "Furniture", "Electronics"],
    "Region": ["North", "South", "North", "East", "West", "North", "East", "West", "South", "North"],
    "Amount": [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

transactions = pd.DataFrame(data)

print(transactions)
```

### Display DataFrame and basic information

```python
print("DataFrame:")
print(transactions)

print("First 5 rows:")
print(transactions.head())

print("Last 5 rows:")
print(transactions.tail())

print("Shape:", transactions.shape)

print("Column names:", transactions.columns)

print("Data types:")
print(transactions.dtypes)
```

#### Display ProductCategory and Amount columns


```python
print(transactions[["ProductCategory", "Amount"]])
```

### Retrieve the last 3 columns

```python
print(transactions.iloc[:, -3:])
```

### Filter transactions from North region with amount above 200

```python
print(transactions[(transactions["Region"] == "North") & (transactions["Amount"] > 200)])
```

### Value counts for the ProductCategory column

```python
print(transactions["ProductCategory"].value_counts())
```

### Find unique regions

```python
print(transactions["Region"].unique())
```

 ### Find the mean amount for each region. 

```python
print(transactions.groupby("Region")["Amount"].mean())
```

### Modify Amount for TransactionID 102 to 165

```python
transactions.loc[transactions["TransactionID"] == 102, "Amount"] = 165

print(transactions)
```

 ### Add Discount column (10% of Amount)

```python
transactions["Discount"] = transactions["Amount"] * 0.10

print(transactions)
```

### Remove the row with TransactionID 109

```python
transactions = transactions[transactions["TransactionID"] != 109]

print(transactions)
```

### Delete Discount column

```python
transactions = transactions.drop("Discount", axis=1)

print(transactions)
```
